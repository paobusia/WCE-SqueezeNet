## Script to replicate frame acquisition from LDPolyp Video Benchmark

```python
import cv2
import os

classes = ['normal']

!mkdir frames

files = os.listdir('./')
files = files[:-2]
print(files)
files = sorted(files,key=lambda x: int(x.split('_')[1][:-4]))
print(files)
```

```python
for f in range(len(files)):
   filename = files[f]
   try:
      os.mkdir(f'frames/{filename[:-4]}')
   except:
      print('folder exists',files[f])
   print(filename)
   cap= cv2.VideoCapture(filename)
   i=0
   if f==24:
      i = 184
   while(cap.isOpened() and i < 500):
     cap.set(cv2.CAP_PROP_POS_MSEC,(i*200))    # added this line
     ret, frame = cap.read()
     if ret == False:
        break
     s2 = str(i+1).zfill(2)
     cv2.imwrite(f'frames/{filename[:-4]}/{filename[:-4]}_{i}.jpg',frame)
     i+=1
   cap.release()

```
